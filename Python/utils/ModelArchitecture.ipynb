```python
import torch
import torchvision.models.segmentation as segmentation
from torchsummary import summary

def build_deeplabv3_resnet101(num_classes=21, pretrained=True):
    """
    DeepLabV3 with ResNet101バックボーンを構築する
    
    Args:
        num_classes (int): セグメンテーションクラス数（デフォルト: 21）
        pretrained (bool): 事前学習済み重みを使用するか（デフォルト: True）
    
    Returns:
        torch.nn.Module: DeepLabV3_resnet101モデル
    """
    model = segmentation.deeplabv3_resnet101(
        pretrained=pretrained,
        num_classes=num_classes
    )
    return model

def print_model_architecture(model, input_size=(3, 224, 224)):
    """
    モデルのアーキテクチャを詳細に表示する
    
    Args:
        model: PyTorchモデル
        input_size (tuple): 入力サイズ (C, H, W)
    """
    print("=" * 80)
    print("DeepLabV3 ResNet101 モデルアーキテクチャ")
    print("=" * 80)
    print("\n[モデル全体の構造]")
    print(model)
    print("\n" + "=" * 80)
    
    print("\n[モデルのサマリー（パラメータ数、FLOPs等）]")
    try:
        summary(model, input_size=input_size, device="cpu")
    except Exception as e:
        print(f"torchinfoによるサマリー表示でエラーが発生しました: {e}")
        print("モデルの基本情報のみ表示します。")
    
    print("\n" + "=" * 80)
    print("\n[モデルのパラメータ統計]")
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"総パラメータ数: {total_params:,}")
    print(f"学習可能パラメータ数: {trainable_params:,}")
    print(f"固定パラメータ数: {total_params - trainable_params:,}")
    
    print("\n" + "=" * 80)
    print("\n[モデルの主要コンポーネント]")
    print(f"バックボーン: {type(model.backbone).__name__}")
    print(f"分類器: {type(model.classifier).__name__}")
    if hasattr(model, 'aux_classifier') and model.aux_classifier is not None:
        print(f"補助分類器: {type(model.aux_classifier).__name__}")
    
    print("\n" + "=" * 80)

if __name__ == "__main__":
    # モデルの構築
    print("DeepLabV3 ResNet101モデルを構築中...")
    model = build_deeplabv3_resnet101(num_classes=21, pretrained=True)
    
    # モデルを評価モードに設定
    model.eval()
    
    # アーキテクチャの表示
    print_model_architecture(model, input_size=(3, 224, 224))
    print(model.backbone.layer4[-1])
```
